```python
# question 1
import pandas as pd
data = {
    "Name": ["A", "B", "C", "D", "E"],
    "Age": [20, 25, 22, 30, 28],
    "Marks": [85, 72, 90, 65, 88],
    "City": ["Delhi", "Mumbai", "Delhi", "Pune", "Mumbai"]
}
df = pd.DataFrame(data)

# Q1
print(df[:3])

# Q2 
print( "\n", df[2:4])

# Q3
print("\n" ,df.shape)

print("\n" ,df.columns)
print("\n" , df.dtypes)
print("\n" , df[["Name" ,"Marks"]])
print("\n" , (df[df["Marks"] >80]) )
print("\n" , df.sort_values(by = "Marks" , ascending = False))
```

```python
#question 2 
import pandas as pd
data = {
    "Age": [20, 21, None, 24, None, 30],
    "Salary": [30000, None, 40000, 50000, None, 60000],
    "Experience": [1, 2, 3, None, 5, 6],
    "Department": ["CS", "IT", "CS", None, "IT", "CS"]
}
df = pd.DataFrame(data)
#Q1
print(df)
#Q2
print("\n" , df.isnull().sum())
# Q3
df["Age"] = df["Age"].fillna(df["Age"].median())
print("\n" , df["Age"])
# Q4
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
print("\n" ,df["Salary"] )
# Q5
df["Experience"] = df["Experience"].fillna(df["Experience"].median())
print("\n" ,df["Experience"] )
# Q6
df["Department"] = df["Department"].fillna(df["Department"].mode()[0])
print("\n" ,df["Department"] )
# Q7
print(df.isnull().sum())
# Q8
print("\n  " , df)
```

```python
#question 3
import pandas as pd
data = {
    "Name": ["A", "B", "C", "D", "E", "F"],
    "Math": [80, 60, 90, 70, 85, 55],
    "Science": [75, 65, 95, 72, 80, 60],
    "English": [85, 70, 88, 75, 90, 50]
}
df = pd.DataFrame(data)

lol = ['Math', 'Science', 'English']

# Q1
df['Average'] = df[lol].mean(axis=1)
print(df["Average"])

df['Total'] = df[lol].sum(axis=1)
print("\n" ,df["Total"])

df['Passed'] = df['Average'] >= 60
print("\n" ,df["Passed"])

highest_student = df.loc[df['Average'].idxmax()]
print( "\n" ,highest_student)

print(df[df['Average'] >= 80])

df_sorted = df.sort_values(by='Average', ascending=False)
print("\n" ,df_sorted)

overall_avg = df['Average'].mean()
print("Overall Average:", overall_avg)

print("\n",df[['Name', 'Average', 'Passed']])
```

```python
#question4
import pandas as pd
data = {
    "Employee": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Department": ["CS", "CS", "IT", "IT", "CS", "IT", "HR", "HR"],
    "Salary": [50000, 60000, 55000, 65000, 70000, 75000, 40000, 45000],
    "Experience": [2, 4, 3, 5, 6, 7, 1, 2]
}
df = pd.DataFrame(data)

#Q1
lol= df.groupby('Department')['Salary'].mean()
print(lol)

#Q2
max_salary_dept = df.groupby('Department')['Salary'].max()
print("\n",max_salary_dept)

#Q3
min_salary_dept = df.groupby('Department')['Salary'].min()
print("\n",min_salary_dept)

#Q4
avg_exp_dept = df.groupby('Department')['Experience'].mean()
print("\n",avg_exp_dept)

#Q5
highest_avg_salary_dept = df.groupby('Department')['Salary'].mean().idxmax()
print("\n","Department with highest average salary:", highest_avg_salary_dept)

#Q6
damn = df.groupby('Department')['Department'].count()
print("\n",damn)

#Q7
sorted = df.groupby('Department')['Salary'].mean().sort_values(ascending=False)
print("\n",sorted)
```
