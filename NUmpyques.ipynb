```python
# question2
import numpy as np
X = np.array([
    [10, 100, 1000],
    [20, 200, 2000],
    [30, 300, 3000],
    [40, 400, 4000],
    [50, 500, 5000]
])

# Q1
lol = np.mean(X , axis = 0)
print("YOUR MEAN is  ")
print (lol)

# Q2
damn = np.std(X , axis = 0)
print ('YOUR STD IS ')
print(damn)

# Q3/4
X_scaled = (X - lol)/damn
print("\n YOUR STADARISED THING IS : " ,X_scaled)

# Q5
new1 = np.mean(X_scaled , axis = 0)
print("YOUR NEW MEAN IS\n " , new1)

new2 = np.std(X_scaled , axis=0)
print("YOUR NEW STD IS \n" ,new2)

print("\n hence verified")
```
